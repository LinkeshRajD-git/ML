```python
# ============================================================
# 1. IMPORTS + READ THE CSV + EDA
# ============================================================
import glob
import time
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import ttest_rel

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.decomposition import PCA
from sklearn.svm import SVR
from sklearn.model_selection import KFold, GridSearchCV
from sklearn.metrics import (
    mean_squared_error, mean_absolute_error, r2_score,
    mean_absolute_percentage_error
)

# Simple CSV selection: put your teacher's CSV beside this notebook.

df = pd.read_csv("sample_machine_rul_dataset.csv")
TARGET = "Remaining_Useful_Life_Hours"  # Change if your target has another exact name

print("Shape:", df.shape)
display(df.head())
print("\nData types:")
df.info()
print("\nMissing values:")
display(df.isnull().sum().to_frame("Missing values"))
print("\nDescriptive statistics:")
display(df.describe(include="all").T)
print("\nDuplicate rows:", df.duplicated().sum())

assert TARGET in df.columns, f"Target not found. Available columns: {list(df.columns)}"

# Remove duplicates and rows without a target (target cannot be imputed for supervised learning)
df = df.drop_duplicates().dropna(subset=[TARGET]).copy()
df[TARGET] = pd.to_numeric(df[TARGET], errors="coerce")
df = df.dropna(subset=[TARGET]).reset_index(drop=True)
y = df[TARGET].copy()

# Remove non-predictive identifier (QP specifies Record ID)
id_cols = [c for c in df.columns if c.lower() in ["record_id", "recordid", "id"]]
# Also remove columns that look like IDs and are almost unique
for c in df.columns:
    if c != TARGET and "id" in c.lower() and df[c].nunique(dropna=True) >= 0.95 * len(df):
        if c not in id_cols:
            id_cols.append(c)
print("\nIdentifier columns removed:", id_cols)

X = df.drop(columns=[TARGET] + [c for c in id_cols if c in df.columns]).copy()
num_cols = X.select_dtypes(include="number").columns.tolist()
cat_cols = X.select_dtypes(exclude="number").columns.tolist()
print("Numeric predictors:", num_cols)
print("Categorical predictors:", cat_cols)

# Target and predictor distributions
plt.figure(figsize=(7, 4))
sns.histplot(y, kde=True)
plt.title("Target distribution")
plt.xlabel(TARGET)
plt.tight_layout()
plt.show()

print("Numeric predictor skewness:")
display(X[num_cols].skew().sort_values(ascending=False).to_frame("Skewness"))

if num_cols:
    X[num_cols].hist(figsize=(12, 8), bins=20, edgecolor="black")
    plt.suptitle("Numeric predictor distributions")
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(max(8, len(num_cols) * 0.7), 5))
    sns.heatmap(X[num_cols].corr(), cmap="coolwarm", center=0)
    plt.title("Correlation between numeric predictors")
    plt.tight_layout()
    plt.show()

    # Outlier analysis only: do not blindly delete potentially valid machine readings.
    n = min(len(num_cols), 8)
    fig, axes = plt.subplots(n, 1, figsize=(9, 2.4 * n))
    if n == 1:
        axes = [axes]
    for ax, col in zip(axes, num_cols[:n]):
        sns.boxplot(x=X[col], ax=ax)
        ax.set_title("Outlier check: " + col)
    plt.tight_layout()
    plt.show()

if cat_cols:
    for col in cat_cols:
        print("\nValue counts:", col)
        display(X[col].value_counts(dropna=False).head(10))

# ============================================================
# 2. LEAKAGE-FREE PREPROCESSING + FULL CHEATSHEET GRID
# ============================================================
# Numeric: median imputation -> StandardScaler
# Categorical: most-frequent imputation -> OneHotEncoder
# PCA: only on scaled numeric features, matching the cheatsheet.
# If old sklearn errors on sparse_output, replace it with sparse=False.
cat_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])
num_pipe_no_pca = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
num_pipe_pca = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("pca", PCA(n_components=0.95))
])

prep_no_pca = ColumnTransformer([
    ("num", num_pipe_no_pca, num_cols),
    ("cat", cat_pipe, cat_cols)
])
prep_pca = ColumnTransformer([
    ("num", num_pipe_pca, num_cols),
    ("cat", cat_pipe, cat_cols)
])

model_no_pca = Pipeline([
    ("prep", prep_no_pca),
    ("svr", SVR(kernel="rbf"))
])
model_pca = Pipeline([
    ("prep", prep_pca),
    ("svr", SVR(kernel="rbf"))
])

# FULL SVR(RBF) grid exactly as on cheatsheet page 5:
# C=[0.1, 1, 10, 100], gamma=['scale', 0.01, 0.1], epsilon=[0.01, 0.1, 0.5]
# Total = 4 * 3 * 3 = 36 parameter combinations per search.
param_grid = {
    "svr__C": [0.1, 1, 10, 100],
    "svr__gamma": ["scale", 0.01, 0.1],
    "svr__epsilon": [0.01, 0.1, 0.5]
}
print("\nHyperparameter combinations per search:",
      len(param_grid["svr__C"]) * len(param_grid["svr__gamma"]) * len(param_grid["svr__epsilon"]))

# Definitions for viva:
# C = penalty/regularization trade-off; gamma = RBF locality; epsilon = no-penalty tube width.
# Search score is negative RMSE because sklearn maximizes scores; negate it to report RMSE.

# ============================================================
# 3. SAME FIVE OUTER FOLDS + INNER GRID SEARCH
# ============================================================
outer_cv = KFold(n_splits=5, shuffle=True, random_state=42)
inner_cv = KFold(n_splits=3, shuffle=True, random_state=42)

def get_metrics(y_true, pred):
    yt = np.asarray(y_true)
    yp = np.asarray(pred)
    rmse = np.sqrt(mean_squared_error(yt, yp))
    mae = mean_absolute_error(yt, yp)
    r2 = r2_score(yt, yp)
    # MAPE is unstable for zero/near-zero actual targets; calculate on non-near-zero rows.
    valid = np.abs(yt) > 1e-8
    mape = 100 * mean_absolute_percentage_error(yt[valid], yp[valid]) if valid.any() else np.nan
    return {"RMSE": rmse, "MAE": mae, "R2": r2, "MAPE_pct": mape}

rows = []
best_params = {"Without PCA": [], "With PCA": []}
fit_seconds = {"Without PCA": [], "With PCA": []}
fold_models = {"Without PCA": [], "With PCA": []}
fold_validation_predictions = {}

for fold, (train_idx, val_idx) in enumerate(outer_cv.split(X, y), start=1):
    X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
    y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]
    row = {"Fold": fold}

    for name, pipe in [("Without PCA", model_no_pca), ("With PCA", model_pca)]:
        search = GridSearchCV(
            estimator=pipe,
            param_grid=param_grid,
            cv=inner_cv,
            scoring="neg_root_mean_squared_error",
            n_jobs=-1,
            refit=True
        )
        start = time.time()
        search.fit(X_train, y_train)
        elapsed = time.time() - start

        fitted = search.best_estimator_
        pred_train = fitted.predict(X_train)
        pred_val = fitted.predict(X_val)
        train_rmse = np.sqrt(mean_squared_error(y_train, pred_train))
        val_metrics = get_metrics(y_val, pred_val)

        row[name + " Train RMSE"] = train_rmse
        for metric, value in val_metrics.items():
            row[name + " " + metric] = value
        row[name + " Best Parameters"] = search.best_params_
        row[name + " Inner CV RMSE"] = -search.best_score_
        row[name + " Time (s)"] = elapsed

        best_params[name].append(search.best_params_)
        fit_seconds[name].append(elapsed)
        fold_models[name].append(fitted)
        fold_validation_predictions[(fold, name)] = (y_val.to_numpy(), pred_val)

    rows.append(row)

results = pd.DataFrame(rows)
results["RMSE Difference (PCA - No PCA)"] = results["With PCA RMSE"] - results["Without PCA RMSE"]

print("\nFold-wise validation RMSE:")
display(results[["Fold", "Without PCA RMSE", "With PCA RMSE",
                 "RMSE Difference (PCA - No PCA)"]])

print("\nMetric summary (mean and standard deviation across outer folds):")
metric_columns = [c for c in results.columns if any(
    tag in c for tag in [" RMSE", " MAE", " R2", " MAPE_pct"]
) and "Difference" not in c and "Best" not in c and "Inner" not in c]
display(results[metric_columns].agg(["mean", "std"]).T)

print("\nTrain-validation RMSE gap (validation RMSE - training RMSE):")
print("Without PCA:", (results["Without PCA RMSE"] - results["Without PCA Train RMSE"]).mean())
print("With PCA:", (results["With PCA RMSE"] - results["With PCA Train RMSE"]).mean())

print("\nBest hyperparameters by fold:")
for name in best_params:
    print("\n", name)
    for fold, params in enumerate(best_params[name], 1):
        print("Fold", fold, params)

print("\nMean fitting+tuning time per outer fold:")
for name in fit_seconds:
    print(name, round(np.mean(fit_seconds[name]), 2), "seconds")

# ============================================================
# 4. PCA VARIANCE REPORT (DESCRIPTIVE) + PLOTS
# ============================================================
# This calculation is only for reporting component count/variance.
# Actual model PCA is still fitted within each training fold in the pipeline.
if num_cols:
    report_numeric = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]).fit_transform(X[num_cols])
    pca_report = PCA().fit(report_numeric)
    cumulative_variance = np.cumsum(pca_report.explained_variance_ratio_)
    n95 = int(np.argmax(cumulative_variance >= 0.95) + 1)
    print(f"\nNumeric dimensions: {report_numeric.shape[1]} -> {n95} components for >=95% variance")
    print(f"Cumulative variance at {n95} components: {cumulative_variance[n95-1]:.4f}")

    plt.figure(figsize=(7, 4))
    plt.plot(range(1, len(cumulative_variance) + 1), cumulative_variance, marker="o")
    plt.axhline(0.95, linestyle="--", label="95% threshold")
    plt.axvline(n95, linestyle=":", label=f"{n95} components")
    plt.xlabel("Number of principal components")
    plt.ylabel("Cumulative explained variance")
    plt.title("PCA cumulative explained variance")
    plt.legend()
    plt.tight_layout()
    plt.show()

# Fold-wise RMSE line plot and distribution plot
plot_data = results[["Fold", "Without PCA RMSE", "With PCA RMSE"]].melt(
    id_vars="Fold", var_name="Model", value_name="RMSE"
)
plt.figure(figsize=(7, 4))
sns.lineplot(data=plot_data, x="Fold", y="RMSE", hue="Model", marker="o")
plt.title("Validation RMSE: same five outer folds")
plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 4))
sns.boxplot(data=plot_data, x="Model", y="RMSE")
sns.stripplot(data=plot_data, x="Model", y="RMSE", color="black", size=5)
plt.title("Fold-wise RMSE distribution")
plt.tight_layout()
plt.show()

# Actual-vs-predicted and residual plots using fold 5 validation predictions
for name in ["Without PCA", "With PCA"]:
    actual, pred = fold_validation_predictions[(5, name)]
    low = min(np.min(actual), np.min(pred))
    high = max(np.max(actual), np.max(pred))

    plt.figure(figsize=(5, 5))
    plt.scatter(actual, pred, alpha=0.7)
    plt.plot([low, high], [low, high], "--")
    plt.xlabel("Actual RUL hours")
    plt.ylabel("Predicted RUL hours")
    plt.title(name + ": actual vs predicted (fold 5)")
    plt.tight_layout()
    plt.show()

    residuals = actual - pred
    plt.figure(figsize=(6, 4))
    plt.scatter(pred, residuals, alpha=0.7)
    plt.axhline(0, linestyle="--")
    plt.xlabel("Predicted RUL hours")
    plt.ylabel("Residual (actual - predicted)")
    plt.title(name + ": residual plot (fold 5)")
    plt.tight_layout()
    plt.show()

# ============================================================
# 5. TWO-SIDED PAIRED T-TEST + EXAM CONCLUSION
# ============================================================
# d = RMSE_PCA - RMSE_noPCA; positive d means PCA RMSE is worse.
t_stat, p_value = ttest_rel(results["With PCA RMSE"], results["Without PCA RMSE"])
mean_diff = results["RMSE Difference (PCA - No PCA)"].mean()

print("\nH0: mean(d) = 0")
print("H1: mean(d) != 0")
print(f"t-statistic = {t_stat:.4f}")
print(f"p-value = {p_value:.6f}")
print("alpha = 0.05")
print(f"Mean paired difference (PCA - no PCA) = {mean_diff:.4f}")

if p_value < 0.05:
    print("Decision: Reject H0. The paired fold RMSE difference is statistically significant.")
else:
    print("Decision: Fail to reject H0. There is insufficient evidence of a significant difference.")

mean_no_pca = results["Without PCA RMSE"].mean()
mean_pca = results["With PCA RMSE"].mean()
print("\nMean validation RMSE without PCA:", round(mean_no_pca, 4))
print("Mean validation RMSE with PCA:", round(mean_pca, 4))
if mean_pca < mean_no_pca:
    print("Lower mean RMSE: With PCA.")
else:
    print("Lower mean RMSE: Without PCA.")

print("""
WRITE-UP GUIDANCE:
1. Report the five fold-wise RMSE values for both models.
2. Report mean +/- standard deviation of RMSE and mean MAE, R2 and MAPE.
3. Compare train-validation RMSE gaps; a large positive gap can indicate overfitting.
4. PCA may reduce numeric dimensions/redundancy but can remove predictive information.
5. State the paired t-test decision using p < 0.05; do not say models are identical when p >= 0.05.
6. Limitations: only 1,000 rows, only five outer folds, t-test assumptions, MAPE near zero.
7. Improvements: broader search, repeated nested CV/independent test set, domain-informed features,
   and carefully justified robust outlier treatment.
""")
```
